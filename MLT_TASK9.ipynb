## Task 9: Perceptron and MLP for Classification

### I. Implementing a Perceptron for Logic Gates

A Perceptron is the simplest form of a neural network, used for binary classification problems. It consists of:
*   Inputs ($x_l, x_2, ..., x_n$)
*   Weights ($w_l, w_2, ..., w_n$)
*   Summation Function: Computes weighted sum
*   Activation Function (Step Function)

Let's implement a perceptron to simulate basic logic gates (AND, OR, NAND).

```python
import numpy as np

# Step activation function
def step_function(x):
    return 1 if x >= 0 else 0

# Perceptron function
def perceptron(inputs, weights, bias):
    weighted_sum = np.dot(inputs, weights) + bias
    return step_function(weighted_sum)

# Logic gate implementation
def logic_gate(gate, X, W, b):
    print(f'\n{gate} Gate Implementation:')
    for x in X:
        output = perceptron(x, W, b)
        print(f'Input: {x}, Output: {output}')

# Define Inputs (Truth Table)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

# Define Weights and Bias for AND, OR, NAND gates
AND_W, AND_b = np.array([1, 1]), -1.5
OR_W, OR_b = np.array([1, 1]), -0.5
NAND_W, NAND_b = np.array([-1, -1]), 1.5

# Run logic gates
logic_gate("AND", X, AND_W, AND_b)
logic_gate("OR", X, OR_W, OR_b)
logic_gate("NAND", X, NAND_W, NAND_b)
```

### II. Implementing a Multi-Layer Perceptron (MLP) from Scratch

A Multi-Layer Perceptron (MLP) is an extension of a perceptron that can handle non-linearly separable problems using:
*   Input layer
*   One or more hidden layers
*   Output layer
*   Backpropagation for training

We will train an MLP on a binary classification problem (classifying points in 2D space) using the backpropagation algorithm.

```python
import numpy as np

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Generate dummy dataset (2D binary classification)
np.random.seed(42)
X = np.random.rand(100, 2) # 100 samples, 2 features
y = ((X[:, 0] + X[:, 1]) > 1).astype(int).reshape(100, 1) # Label: 1 if sum > 1, else 0

# Initialize MLP parameters
input_size = 2
hidden_size = 4
output_size = 1
lr = 0.1 # Learning rate

# Weights and biases
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Training MLP using Backpropagation
epochs = 1000
for epoch in range(epochs):
    # Forward pass
    hidden_layer_input = np.dot(X, W1) + b1
    hidden_layer_output = sigmoid(hidden_layer_input)
    output_layer_input = np.dot(hidden_layer_output, W2) + b2
    output_layer_output = sigmoid(output_layer_input)

    # Compute loss (Mean Squared Error)
    loss = np.mean((output_layer_output - y) ** 2)

    # Backpropagation
    output_error = output_layer_output - y
    output_delta = output_error * sigmoid_derivative(output_layer_output)

    hidden_error = output_delta.dot(W2.T)
    hidden_delta = hidden_error * sigmoid_derivative(hidden_layer_output)

    # Update weights and biases
    W2 -= hidden_layer_output.T.dot(output_delta) * lr
    b2 -= np.sum(output_delta, axis=0, keepdims=True) * lr
    W1 -= X.T.dot(hidden_delta) * lr
    b1 -= np.sum(hidden_delta, axis=0, keepdims=True) * lr

    # Print loss every 100 epochs
    if epoch % 100 == 0:
        print(f'Epoch {epoch}, Loss: {loss:.4f}')

# Test Model
final_hidden_layer_output = sigmoid(np.dot(X, W1) + b1)
final_predictions = sigmoid(np.dot(final_hidden_layer_output, W2) + b2)
predictions = (final_predictions > 0.5).astype(int)
accuracy = np.mean(predictions == y) * 100

print(f'\nFinal Training Accuracy: {accuracy:.2f}%')
```
