# Back Propagation Neural Network in Image Data set

Backpropagation Neural Network (BPNN) on image data for feature extraction and classification using TensorFlow/Keras. This example demonstrates using a Convolutional Neural Network (CNN), which is a specialized ANN optimized for image data.

**Procedure:**
1.  **Load Image Dataset:** Use a dataset like MNIST (handwritten digits) or CIFAR-10 (object recognition).
2.  **Preprocess the Data:**
    *   Normalize pixel values (0-255 -> 0-1).
    *   Reshape for compatibility with neural networks.
3.  **Build the Neural Network Model:**
    *   Use Convolutional Layers for feature extraction.
    *   Use Pooling Layers to reduce spatial dimensions.
    *   Use Fully Connected Layers to classify extracted features.
4.  **Compile the Model:**
    *   Use Categorical Crossentropy Loss for multi-class classification.
    *   Use Adam Optimizer to update weights via backpropagation.
5.  **Train the Model:**
    *   Split the dataset into training and validation sets.
    *   Train the model using backpropagation with multiple epochs.
6.  **Evaluate Performance:**
    *   Check Accuracy, Loss, and Confusion Matrix.
    *   Visualize results with sample predictions.

```python
# Import necessary libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
```

### 1. Load and Preprocess the Data

```python
# Load MNIST dataset (handwritten digits)
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalize pixel values (0 to 255 -> 0 to 1)
X_train, X_test = X_train / 255.0, X_test / 255.0
```

### 2. Reshape Data and One-Hot Encode Labels

```python
# Flatten 28x28 images to 1D array of 784 pixels
X_train = X_train.reshape(-1, 784)
X_test = X_test.reshape(-1, 784)

# Convert labels to one-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

### 3. Build the Artificial Neural Network (ANN) Model

```python
model = Sequential([
    Dense(128, activation='relu', input_shape=(784,)), # First Hidden Layer
    Dense(64, activation='relu'),   # Second Hidden Layer
    Dense(10, activation='softmax')     # Output Layer (10 classes)
])
```

### 4. Compile the Model

```python
# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

### 5. Train the Model

```python
# Train the model using backpropagation
model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))
```

### 6. Evaluate Performance

```python
# Evaluate the model on test data
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f'\nTest Accuracy: {test_acc * 100:.2f}%')
```

The model correctly classifies about 97.15% of handwritten digits (this value might vary slightly due to random initialization or different runs).
